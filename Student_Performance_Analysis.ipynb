```python

# Module 5 - Student Performance Analysis

# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Student Performance Dataset
url = "https://raw.githubusercontent.com/MakroMB/student-performance-analysis/main/data/student-mat.csv"

df = pd.read_csv(url, sep=";")

# Display dataset
print("Dataset loaded successfully!")

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
display(df.head())
```

```python

# Step 2 - Understanding the Dataset

# Display column names
print("Column Names:")
print(df.columns.tolist())

# Display dataset information
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Statistical summary
print("\nStatistical Summary:")
display(df.describe())

```

```python

# Step 3 - Data Cleaning and Preparation

# Check duplicate records
print("Duplicate Records:", df.duplicated().sum())

# Check missing values
print("\nMissing Values:", df.isnull().sum().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Display final dataset shape
print("\nDataset Shape After Cleaning:", df.shape)

print("\nData Cleaning Completed Successfully!")
```

```python

# Step 4 - Exploratory Data Analysis

# Average student grades
print("Average Grade 1:", df['G1'].mean())
print("Average Grade 2:", df['G2'].mean())
print("Average Final Grade:", df['G3'].mean())

# Highest and lowest final grades
print("\nHighest Final Grade:", df['G3'].max())
print("Lowest Final Grade:", df['G3'].min())

# Overall student performance
print("\nStudent Performance Analysis:")
print(df[['G1', 'G2', 'G3']].describe())
```

```python

# Step 5 - Student Performance Visualization

import matplotlib.pyplot as plt
import seaborn as sns

# 1. Distribution of Final Grades
plt.figure(figsize=(8,5))
sns.histplot(df['G3'], bins=10, kde=True)
plt.title("Distribution of Final Student Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.show()

# 2. Comparison of Average Grades
plt.figure(figsize=(8,5))
avg_grades = df[['G1', 'G2', 'G3']].mean()
avg_grades.plot(kind='bar')
plt.title("Average Student Grades")
plt.xlabel("Exam")
plt.ylabel("Average Marks")
plt.show()

# 3. Study Time vs Final Grade
plt.figure(figsize=(8,5))
sns.scatterplot(x='studytime', y='G3', data=df)
plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Final Grade")
plt.show()
```

```python

# Step 6 - Correlation Analysis

# Calculate correlation
correlation = df[['studytime', 'failures', 'absences',
                  'G1', 'G2', 'G3']].corr()

print("Correlation with Final Grade (G3):")
print(correlation['G3'].sort_values(ascending=False))

# Create correlation heatmap
plt.figure(figsize=(9,6))
sns.heatmap(correlation, annot=True, cmap='coolwarm',
            fmt='.2f', linewidths=0.5)

plt.title("Student Performance Correlation Heatmap")
plt.show()
```

```python

# Step 7 - Student Performance Classification

def classify_grade(grade):
    if grade >= 15:
        return "High Performance"
    elif grade >= 10:
        return "Average Performance"
    else:
        return "Low Performance"

# Create a new column
df['Performance'] = df['G3'].apply(classify_grade)

# Display student performance
print("Student Performance Categories:")
print(df['Performance'].value_counts())

# Visualize performance categories
plt.figure(figsize=(8,5))
sns.countplot(x='Performance', data=df)
plt.title("Student Performance Classification")
plt.xlabel("Performance Category")
plt.ylabel("Number of Students")
plt.show()
```
